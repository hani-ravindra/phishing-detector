```python
# Cell 1: Import Libraries
import pandas as pd

print("Libraries imported.")
```

```python
# Cell 2: Load and Process the Kaggle Dataset
print("--- Processing Kaggle Dataset (phishing_site_urls.csv) ---")

# Load the data
kaggle_df = pd.read_csv('data/phishing_site_urls.csv')

# Standardize the labels: 'bad' -> 1 (phishing), 'good' -> 0 (legitimate)
kaggle_df['Label'] = kaggle_df['Label'].map({'bad': 1, 'good': 0})

# Keep only the essential columns
kaggle_df = kaggle_df[['URL', 'Label']]

print(f"Kaggle data processed. Shape: {kaggle_df.shape}")
print("Label distribution:")
print(kaggle_df['Label'].value_counts())
kaggle_df.head()
```

```python
# Cell 3: Load and Process the Tranco Top Sites Dataset
print("\n--- Processing Tranco Top 1 Million Dataset (top-1m.csv) ---")

# Load the data (let's take the top 200,000 for a large, manageable sample)
tranco_df = pd.read_csv('data/top-1m.csv', header=None, names=['Rank', 'Domain'])
tranco_df = tranco_df.head(200000)

# Convert domain names to full URLs
tranco_df['URL'] = 'http://www.' + tranco_df['Domain']

# Label all these sites as legitimate (0)
tranco_df['Label'] = 0

# Keep only the essential columns
tranco_df = tranco_df[['URL', 'Label']]

print(f"Tranco data processed. Shape: {tranco_df.shape}")
tranco_df.head()
```

```python
# Cell 4: Combine and Clean the Datasets (UPDATED)
print("\n--- Combining Kaggle and Tranco Datasets ---")

# Use pd.concat to merge the two dataframes
combined_df = pd.concat([kaggle_df, tranco_df], ignore_index=True)

# Remove any duplicates that might have been introduced
combined_df.drop_duplicates(subset=['URL'], inplace=True)

# --- THE FIX IS HERE ---
# Drop any rows where the 'Label' is missing (NaN) to prevent errors
combined_df.dropna(subset=['Label'], inplace=True)

# As good practice, ensure the Label column is an integer type (0 or 1)
combined_df['Label'] = combined_df['Label'].astype(int)
# --- END FIX ---

# Shuffle the dataset to ensure randomness
combined_df = combined_df.sample(frac=1).reset_index(drop=True)


print(f"Final combined dataset shape: {combined_df.shape}")
print("Final label distribution:")
print(combined_df['Label'].value_counts())
```

```python
# Cell 5: Save the Final Combined Dataset
print("\n--- Saving the final dataset ---")

# Save to a new file in the data folder
output_path = 'data/combined_dataset.csv'
combined_df.to_csv(output_path, index=False)

print(f"Successfully saved the combined dataset to '{output_path}'")
print("This file is now ready to be used by the '1_Data_Preparation.ipynb' notebook.")
```
